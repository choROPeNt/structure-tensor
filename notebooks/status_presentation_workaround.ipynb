```python
import os
from pathlib import Path
import h5py 

import numpy as np
import nrrd

import matplotlib.pyplot as plt

from structure_tensor.h5_io import load_h5_datasets
```

```python
root_dir = Path("/Volumes/projects/3dseg/data/BIIAX/285_10-layer/03/285_10-layer_03_0")

nrrd_path = root_dir /  "BIIAX_196(245)_02_tomo-A_recon_crop_ztrim150.seg_.nrrd"
file_path = root_dir / "285_10-layer_03_0.vol.seg.h5"
```

```python
data , meta = nrrd.read(nrrd_path)
print(data.shape)
print(meta)
```

```python
key = "seg_"
with h5py.File(file_path, "a") as f:
    if key in f:
        del f[key]
    f.create_dataset(key, data=data, compression="gzip", compression_opts=4)
    print(f"Written '{key}' {data.shape} {data.dtype} → {file_path.name}")
```
